#RQ 1

Initially, a one-way ANOVA was used to compare the mean financial exclusion scores between developing and underdeveloped economies. Although a two-sample independent t-test is sufficient when comparing the means of two groups, one-way ANOVA was selected because it provides an equivalent test of mean differences while offering greater flexibility for extending the analysis to multiple groups. Furthermore, with a large sample size, both the independent t-test and one-way ANOVA produce identical conclusions for comparisons involving two groups.

##One Way ANOVA Test

### RQ1. Is there a significant difference in the level of financial exclusion between underdeveloped and developing economies?

We will test the null hypothesis

>$H_0$: All the population variances are equal

against the alternative hypothesis

>$H_a$: At least one variance is different from the rest

```python
df['fin_excl_score_wgt'] = df['fin_excl_score'] * df['wgt']
```

```python
print(df.groupby('economy_type')['fin_excl_score_wgt'].mean())

fig, ax = plt.subplots(figsize = (6,6))
a = sns.boxplot(x= "economy_type", y = 'fin_excl_score_wgt' , data = df, hue = 'economy_type')
a.set_title("Financial Exclusion w.r.t. Economy Type)", fontsize=15)
plt.show()
```

Now, the normality and equality of variance assumptions need to be checked.

* For testing of normality, Shapiro-Wilk’s test is applied to the response variable.

* For equality of variance, Levene test is applied to the response variable.

### Shapiro-Wilk’s test

We will test the null hypothesis

>$H_0:$ Financial Exclusion follows a normal distribution against the alternative hypothesis

>$H_a:$ Financial Exclusion does not follow a normal distribution

```python
# Assumption 1: Normality
# import the required function
from scipy import stats

# find the p-value
w, p_value = stats.shapiro(df['fin_excl_score_wgt'])
print('The p-value is', p_value)
```

Since p-value of the test is very small, we will reject the null hypothesis that the response follows the normal distribution.

### Levene’s test

We will test the null hypothesis

>$H_0$: All the population variances are equal

against the alternative hypothesis

>$H_a$: At least one variance is different from the rest

```python
#Assumption 2: Homogeneity of Variance
#import the required function
from scipy.stats import levene
statistic, p_value = levene( df['fin_excl_score_wgt'][df['economy_type']=="developing economy"],
                                   df['fin_excl_score_wgt'][df['economy_type']=="underdeveloped economy"])
# find the p-value
print('The p-value is', p_value)
```

Since the p-value is small, we will reject the null hypothesis of homogeneity of variances.

Since both the assmption are not satisfied we will move on to Two Sample Independent T test, and drop the variable specifically created for test.

```python
df = df.drop(columns=['fin_excl_score_wgt'])
```

## Two Sample Independent T test

### Hypotheses for Two Sample Independent T-test

- **Null Hypothesis ($H_0$):** There is no significant difference in the mean financial exclusion score between underdeveloped and developing economies.
  $H_0: \mu_{\text{underdeveloped}} = \mu_{\text{developing}}$

- **Alternative Hypothesis ($H_1$):** There is a significant difference in the mean financial exclusion score between underdeveloped and developing economies.
  $H_1: \mu_{\text{underdeveloped}} \neq \mu_{\text{developing}}$

```python
from statsmodels.stats.weightstats import ttest_ind

# Separate the financial exclusion scores and weights for each economy type
underdeveloped_economy_scores = df[df['economy_type'] == 'underdeveloped economy']['fin_excl_score']
underdeveloped_economy_weights = df[df['economy_type'] == 'underdeveloped economy']['wgt']

developing_economy_scores = df[df['economy_type'] == 'developing economy']['fin_excl_score']
developing_economy_weights = df[df['economy_type'] == 'developing economy']['wgt']

# Perform the independent two-sample t-test with weights
t_statistic, p_value, _ = ttest_ind(underdeveloped_economy_scores, developing_economy_scores,
                                    weights=(underdeveloped_economy_weights, developing_economy_weights))

print(f"Weighted T-statistic: {t_statistic:.4f}")
print(f"Weighted P-value: {p_value:.4e}")
```

### Interpretation of Results

The weighted two-sample t-test has been successfully executed. The results show a Weighted T-statistic of 47.2083 and a Weighted P-value of 0.0000e+00. Since the P-value is extremely small (effectively zero) and much less than the typical significance level of 0.05, we reject the null hypothesis. This means there is a statistically significant difference in the mean financial exclusion score between underdeveloped and developing economies. The positive T-statistic suggests that the financial exclusion score is significantly higher in underdeveloped economies.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np # Import numpy for np.average

def plot_financial_exclusion_by_category(df, column, labels_map=None, weight_column=None):
    """
    Plots the average financial exclusion score for a categorical variable.

    Parameters
    ----------
    df : pandas.DataFrame
        DataFrame containing the data.
    column : str
        Name of the categorical column to plot.
    labels_map : dict, optional
        Dictionary containing category label mappings for readability.
    weight_column : str, optional
        Name of the column to use as weights for calculating the average.
    """

    # Calculate average financial exclusion score
    if weight_column:
        df_plot = (
            df.groupby(column).apply(lambda x: np.average(x["fin_excl_score"], weights=x[weight_column]), include_groups=False)
            .reset_index(name="fin_excl_score")
        )
    else:
        df_plot = (
            df.groupby(column)["fin_excl_score"]
            .mean()
            .reset_index()
        )

    # Replace numerical values with descriptive labels if available
    if labels_map and column in labels_map:
        df_plot[column] = df_plot[column].map(labels_map[column])

    # Create plot
    plt.figure(figsize=(10, 6))
    ax = sns.barplot(
        data=df_plot,
        x=column,
        y="fin_excl_score",
        palette="viridis",
        hue=column,
        legend=False
    )

    plt.title(f"Financial Exclusion by {column.replace('_', ' ').title()}")
    plt.xlabel(column.replace("_", " ").title())
    plt.ylabel("Average Financial Exclusion Score")
    plt.xticks(rotation=45, ha="right")

    # Display values on bars
    for container in ax.containers:
        ax.bar_label(container, fmt="%.3f")

    plt.tight_layout()
    plt.show()
```

```python
def treat_outliers(df, col):
    """
    treats outliers in a variable
    col: str, name of the numerical variable
    df: dataframe
    col: name of the column
    """
    Q1 = df[col].quantile(0.25)  # 25th quantile
    Q3 = df[col].quantile(0.75)  # 75th quantile
    IQR = Q3 - Q1                # Inter Quantile Range (75th perentile - 25th percentile)
    lower_whisker = Q1 - 1.5 * IQR
    upper_whisker = Q3 + 1.5 * IQR

    # all the values smaller than lower_whisker will be assigned the value of lower_whisker
    # all the values greater than upper_whisker will be assigned the value of upper_whisker
    # the assignment will be done by using the clip function of NumPy
    df[col] = np.clip(df[col], lower_whisker, upper_whisker)

    return df
```

```python
# defining a function to compute different metrics to check performance of a classification model built using sklearn
def model_performance_classification_sklearn(model, predictors, target):
    """
    Function to compute different metrics to check classification model performance

    model: classifier
    predictors: independent variables
    target: dependent variable
    """

    # predicting using the independent variables
    pred = model.predict(predictors)

    acc = accuracy_score(target, pred)  # to compute Accuracy
    recall = recall_score(target, pred)  # to compute Recall
    precision = precision_score(target, pred)  # to compute Precision
    f1 = f1_score(target, pred)  # to compute F1-score

    # creating a dataframe of metrics
    df_perf = pd.DataFrame(
        {"Accuracy": acc, "Recall": recall, "Precision": precision, "F1": f1,},
        index=[0],
    )

    return df_perf
```

```python
def confusion_matrix_sklearn(model, predictors, target):
    """
    To plot the confusion_matrix with percentages

    model: classifier
    predictors: independent variables
    target: dependent variable
    """
    y_pred = model.predict(predictors)
    cm = confusion_matrix(target, y_pred)
    labels = np.asarray(
        [
            ["{0:0.0f}".format(item) + "\n{0:.2%}".format(item / cm.flatten().sum())]
            for item in cm.flatten()
        ]
    ).reshape(2, 2)

    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=labels, fmt="")
    plt.ylabel("True label")
    plt.xlabel("Predicted label")
```

### Hypotheses for Two-Sample Proportion Z-test for 'Receive Wages Other than Bank'

- **Null Hypothesis ($H_0$):** There is no significant difference in the proportion of financially excluded individuals who receive wages 'Other than bank' between developing and underdeveloped economies.
  $H_0: p_{\text{developing_other_bank}} = p_{\text{underdeveloped_other_bank}}$

- **Alternative Hypothesis ($H_1$):** There is a significant difference in the proportion of financially excluded individuals who receive wages 'Other than bank' between developing and underdeveloped economies.
  $H_1: p_{\text{developing_other_bank}} \neq p_{\text{underdeveloped_other_bank}}$

```python
from statsmodels.stats.proportion import proportions_ztest

financially_excluded_df = df[df['fin_excl_score'] >= 0.5]

# Separate data by economy type for financially excluded individuals
developing_economy_fe_wages = financially_excluded_df[financially_excluded_df['economy_type'] == 'developing economy']
underdeveloped_economy_fe_wages = financially_excluded_df[financially_excluded_df['economy_type'] == 'underdeveloped economy']

# Calculate weighted counts of financially excluded individuals who receive wages 'Other than bank'
count_developing_other_bank = (developing_economy_fe_wages['receive_wages'].apply(lambda x: 1 if x == 'Other than bank' else 0) * developing_economy_fe_wages['wgt']).sum()
count_underdeveloped_other_bank = (underdeveloped_economy_fe_wages['receive_wages'].apply(lambda x: 1 if x == 'Other than bank' else 0) * underdeveloped_economy_fe_wages['wgt']).sum()

# Calculate total weighted number of financially excluded individuals for each economy type
nobs_developing_fe = developing_economy_fe_wages['wgt'].sum()
nobs_underdeveloped_fe = underdeveloped_economy_fe_wages['wgt'].sum()

# Perform the two-sample proportion z-test
counts = [count_developing_other_bank, count_underdeveloped_other_bank]
nobs = [nobs_developing_fe, nobs_underdeveloped_fe]

z_statistic, p_value = proportions_ztest(counts, nobs)

print(f"Weighted Proportion of 'receive wages Other than bank' & financially excluded in developing economies: {count_developing_other_bank / nobs_developing_fe:.4f}")
print(f"Weighted Proportion of 'receive wages Other than bank' & financially excluded in underdeveloped economies: {count_underdeveloped_other_bank / nobs_underdeveloped_fe:.4f}")
print(f"\nWeighted Z-statistic: {z_statistic:.4f}")
print(f"Weighted P-value: {p_value:.4e}")
```

### Interpretation of Results

The results indicate that the weighted proportion of financially excluded individuals who receive wages 'Other than bank' is 0.1117 in developing economies and 0.0949 in underdeveloped economies. The weighted Z-statistic is 5.4079, and the P-value is 6.3756e-08. Since the P-value is extremely small (much less than the common significance level of 0.05), we reject the null hypothesis. This means there is a statistically significant difference in the proportion of financially excluded individuals who receive wages 'Other than bank' between developing and underdeveloped economies. The positive Z-statistic suggests that this proportion is significantly higher in developing economies.

- **Null Hypothesis ($H_0$):** There is no significant difference in the proportion of financially excluded individuals associated with connectivity between developing and underdeveloped economies.
  $H_0: p_{\text{developing}} = p_{\text{underdeveloped}}$

- **Alternative Hypothesis ($H_1$):** There is a significant difference in the proportion of financially excluded individuals associated with connectivity between developing and underdeveloped economies.

```python
missing_internet_use = df['internet_use'].isnull().sum()
print(f"Number of missing values in 'internet_use': {missing_internet_use}")
```

```python
from statsmodels.stats.proportion import proportions_ztest

financially_excluded_df = df[df['fin_excl_score'] >= 0.5]

# Separate data by economy type

developing_economy_fe = financially_excluded_df[financially_excluded_df['economy_type'] == 'developing economy']
underdeveloped_economy_fe = financially_excluded_df[financially_excluded_df['economy_type'] == 'underdeveloped economy']

# Calculate weighted counts of financially excluded individuals with internet use for each economy type
# 'internet_use' is 1 for connected, 0 for not connected. Multiply by 'wgt' to apply weights.
count_developing_connected = (developing_economy_fe['internet_use'] * developing_economy_fe['wgt']).sum()
count_underdeveloped_connected = (underdeveloped_economy_fe['internet_use'] * underdeveloped_economy_fe['wgt']).sum()

# Calculate total weighted number of financially excluded individuals for each economy type
nobs_developing_fe = developing_economy_fe['wgt'].sum()
nobs_underdeveloped_fe = underdeveloped_economy_fe['wgt'].sum()

# Perform the two-sample proportion z-test
counts = [count_developing_connected, count_underdeveloped_connected]
nobs = [nobs_developing_fe, nobs_underdeveloped_fe]

z_statistic, p_value = proportions_ztest(counts, nobs)

print(f"Weighted Proportion of connected & financially excluded in developing economies: {count_developing_connected / nobs_developing_fe:.4f}")
print(f"Weighted Proportion of connected & financially excluded in underdeveloped economies: {count_underdeveloped_connected / nobs_underdeveloped_fe:.4f}")
print(f"\nWeighted Z-statistic: {z_statistic:.4f}")
print(f"Weighted P-value: {p_value:.4e}")
```

### Interpretation of Results

With an extremely small P-value (effectively zero), which is far below any typical significance level (e.g., 0.05 or 0.01), we confidently reject the null hypothesis. This means there is a statistically significant difference in the proportion of financially excluded individuals who are connected to the internet between developing and underdeveloped economies.

Specifically, a much higher weighted proportion of financially excluded individuals in developing economies (0.6561) are connected to the internet compared to those in underdeveloped economies (0.2718). The large positive Z-statistic further supports this significant difference.

```python
from scipy.stats import chi2_contingency

# Filter data for underdeveloped economies
underdeveloped_df = df[df['economy_type'] == 'underdeveloped economy'].copy()

# Create a binary 'is_financially_excluded' column based on the new threshold
# True if fin_excl_score > 0.5, False otherwise
underdeveloped_df['is_financially_excluded'] = (underdeveloped_df['fin_excl_score'] >= 0.5).astype(int)

# Create a weighted contingency table between 'internet_use' and 'is_financially_excluded'
# by summing the 'wgt' column for each combination
weighted_contingency_table = underdeveloped_df.groupby(['internet_use', 'is_financially_excluded'])['wgt'].sum().unstack(fill_value=0)

print("Weighted Contingency Table (internet_use vs is_financially_excluded in underdeveloped economies):")
display(weighted_contingency_table)

# Perform the Chi-Square Test of Independence on the weighted contingency table
chi2, p_value, dof, expected = chi2_contingency(weighted_contingency_table)

print(f"\nWeighted Chi-Square Statistic: {chi2:.4f}")
print(f"Weighted P-value: {p_value:.4e}")
print(f"Degrees of Freedom: {dof}")
# print("Expected Frequencies:")
# print(expected)
```

### Interpretation of Results

With an extremely small P-value (1.3359e-56), which is significantly less than the conventional significance levels (e.g., 0.05 or 0.01), we reject the null hypothesis. The null hypothesis stated that there is no association between internet use and financial exclusion in underdeveloped economies.

Therefore, we can conclude that there is a statistically significant association between internet use and financial exclusion within underdeveloped economies. The large Chi-Square statistic further supports this strong association. This implies that internet use is not independent of financial exclusion in these economies, and access to the internet plays a role in the financial inclusion landscape of underdeveloped nations.
