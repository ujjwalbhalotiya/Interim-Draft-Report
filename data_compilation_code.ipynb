```python
columns = [
    'educ', 'economycode', 'pop_adult', 'wgt', 'female', 'age', 'inc_q', 'emp_in',
    'urbanicity', 'account', 'borrowed', 'saved', 'receive_wages',
    'receive_transfers', 'receive_pensions', 'pay_utilities', 'fin19',
    'fin22a', 'fin24a', 'fin24c', 'fh2a', 'fh1', 'fh2',
    'fin37', 'fin38', 'fin42', 'internet_use', 'con1'
]

df_subset = df[columns]
```

```python
# loading the dataset
incomedata = pd.read_excel("CountryIncomeGroup.xlsx")
```

```python
incomedata.head()
```

```python
incomedata_renamed = incomedata.rename(columns={'Code': 'economycode'})

df_subset = pd.merge(df_subset, incomedata_renamed[['economycode', 'Income group']], on='economycode', how='left')

display(df_subset.head())
```

```python
unique_income_groups = df_subset['Income group'].unique()
display(unique_income_groups)
```

```python
df_subset = df_subset[df_subset['Income group'] != 'High income']
display(df_subset.head())
```

```python
economy_counts = df_subset['Income group'].value_counts()
display(economy_counts)
```

```python
df_subset.to_csv('combined_dataset_weight.csv', index=False)
print('combined_dataset_weight.csv has been successfully created.')
```
